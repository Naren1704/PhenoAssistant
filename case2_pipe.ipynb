```python
# compare if case2_task1_wlog = case2 (original case2)
import pandas as pd

def compare_csv(file1, file2):
    df1 = pd.read_csv(file1)
    df2 = pd.read_csv(file2)

    # Check same set of columns
    if set(df1.columns) != set(df2.columns):
        return False

    # Reorder df2 columns to match df1
    df2 = df2[df1.columns]

    return df1.equals(df2)

print(compare_csv('./results/Case2/potato_phenotypes.csv', './results/Case2_Task1_wlog/potato_phenotypes_and_metadata.csv'))
```
