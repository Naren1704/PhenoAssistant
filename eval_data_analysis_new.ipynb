```python
# Task 17 evaluation
import pandas as pd

# Load the CSV file
df = pd.read_csv("./data_for_eval/potato_metadata.csv")

# Filter rows for Voyager
voyager_df = df[df["Variety"].str.lower() == "voyager"]

# Compute mean and standard deviation
mean_dw = voyager_df["manual_dried_weight"].mean()
std_dw = voyager_df["manual_dried_weight"].std()

print(f"Voyager manual dried weight - Mean: {mean_dw}, Std: {std_dw}")
```

```python
# Task 18 evaluation
df = pd.read_csv("./data_for_eval/aracrop_phenotypes.csv")
filtered_df = df[(df['ecotype'] == 'pgm') & (df['days_after_sowing'] == 1)]
mean_count = filtered_df['leaf_count'].mean()
std_count = filtered_df['leaf_count'].std()
print(f"Mean leaf count (ecotype=pgm, day=1): {mean_count}")
print(f"Standard deviation of leaf count: {std_count}")
```
